```python

# =====================================================
# Exploratory Data Analysis (EDA)
# Collaborative Multi-Agent Dynamic Pricing Project
# =====================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load Dataset
df = pd.read_csv("../dataset/ecommerce_pricing.csv")

print("Dataset loaded successfully!")
print("\nDataset Shape:", df.shape)

# 2. Display First Five Rows
print("\nFirst Five Rows:")
display(df.head())

# 3. Dataset Information
print("\nDataset Information:")
df.info()

# 4. Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Statistical Summary
print("\nStatistical Summary:")
display(df.describe())

# 6. Product Category Distribution
print("\nProduct Categories:")
print(df["category"].value_counts())

# 7. Revenue Distribution
plt.figure(figsize=(7, 5))
sns.histplot(df["revenue"], kde=True)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.show()

# 8. Product Price Distribution
plt.figure(figsize=(7, 5))
sns.histplot(df["product_price"], kde=True)
plt.title("Product Price Distribution")
plt.xlabel("Product Price")
plt.show()

# 9. Price vs Units Sold
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="product_price", y="units_sold")
plt.title("Product Price vs Units Sold")
plt.show()

# 10. Correlation Heatmap
plt.figure(figsize=(10, 7))
numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()

print("\nEDA Completed Successfully!")
```
